```python
import tensorflow as tf
mnist = tf.keras.datasets.mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

> MNIST 모델 실습  
단순히 적층만 하면 되는 `Sequential`  
즉, 리스트에 레이어를 순서대로 나열하면, 그 순서대로 연결해주는 방식  

1. 입력 형태를 정한다.
2. 레어어를 붙인다.
3. 레이어를 붙인다.

> 쌓는 방향은 입력부터 출력 방향으로.

- 입력 		→ 레이어1 → 레이어2 	→ 레이어3 → 출력
- (28,28) 	→ Flatten → Dense(64) 	→ Dense(10)

```python
model = tf.keras.models.Sequential(
	[
		tf.keras.Input(shape=(28, 28)), #선언만 하는 것, 진짜 데이터 넣는 것은 나중에.
		# 이제 28*28해서 784개가 될 것. 
		# (Flatten은 값을 바꾸지 않고 모양만 바꿈)
		# 이제 이미지는 784차원 벡터가 되었음.
		# 이때부터는 이미지 픽셀들이 아닌, 그냥 숫자 벡터에 더 가까움

		tf.keras.layers.Flatten(), 
		# 입력 shpae: (784, )
		# 출력 shape: (64, )
		
		tf.keras.layers.Dense(units=64, activation="relu"),
		# Dense(64, activation="relu") = ReLU(xW + b)
		# 가중치 W의 shape은 (784, 64)
		# 편향 b의 shape은 (64, )
		# 선형계산은 xW + b
		# 이후 ReLU로 음수는 0, 양수는 유지.

		tf.keras.layers.Dense( units = 10, activation="softmax")
		# 가중치 W의 shape은 (64, 10)
		# 편향 b의 shape은 (10, )
		# 선형계산은 xW + b
		# 이후 softmax로 10개의 값의 합이 1이 되는 형태로 치환
	]
)
```

```python
model.compile(
	# model.compile()은 3가지를 정한다
	# 1. 무엇을 최소화할 것인가? (loss)
	# 2. 어떻게 최소화할 것인가? (optimizer)
	# 3. 무엇을 모니터링할 것인가? (metrics)

	# 지금은 0~9 정수 중 하나를 고르는 분류 문제.
	loss = "sparse_categorical_crossentropy",
	optimizer = "Adam",
	metrics = ["accuracy"]
)
```

```python
model.fit( X_train, y_train, epochs= 50)
```
