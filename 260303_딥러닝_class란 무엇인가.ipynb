```python
# 그러니까, 필요한 기능들을 모두 함수화를 최대한 하려고 함.
# 기존에 있던 함수? 1명 학생에 대한 정보를 생성하는 함수.
# 새로이 만들 함수? 이를 바탕으로 총점을 구하는 함수
# 또, 한 명 학생에 대한 정보를 바탕으로 평균을 구하는 함수.
# 마지막으로, 모든 정보들을 출력하는 함수
```

```python
# 1명 학생에 대한 정보를 생성하는 함수
def create_student(name, kor, math, eng, sci):
    return {"name" : name, "korean" : kor, "math" : math, "english" : eng, "science" : sci}
```

```python
def student_get_sum(student):
    score_sum = student["korean"] + student["math"] + student["english"] + student["science"]
    return score_sum
```

```python
def student_get_avg(student):
    score_avg = student_get_sum(student)/4
    return score_avg
```

```python
def student_get_sum_avg(student):

    score_sum = student["korean"] + student["math"] + student["english"] + student["science"]

    score_avg = student_get_sum(student)/4

    return score_sum, score_avg
```

```python
def student_to_string(student):
    temp = "{0}\t{1}\t{2}".format(student["name"],
                                  student_get_sum(student),
                                  student_get_avg(student)
                                  )
    return temp
```

```python
test_std = {"name" : "AAA", "korean" : 98, "math" : 93, "english" : 79, "science" : 95}

student_get_sum_avg(test_std)
```

```python
# 우리반 "학생" 별로 4과목 성적에 대한 총점/평균을 출력하고자 함

print("이름", "총점", "평균", sep = "\t")

for student in students:
    print(student_to_string(student))
```
