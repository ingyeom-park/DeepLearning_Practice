```python
# 예: 학생들에 대한 국어, 영어, 수학, 과학 점수를 처리하고자 함.
# 그렇다면 우리반 전체 학생들은 리스트,. 개별 학생들에 대한 과목 정보는 dictionary
```

```python
students = [
    {"name" : "AAA", "korean" : 98, "math" : 93, "english" : 79, "science" : 95},
    {"name" : "BBB", "korean" : 94, "math" : 83, "english" : 74, "science" : 85},
    {"name" : "CCC", "korean" : 92, "math" : 93, "english" : 73, "science" : 75},
    {"name" : "DDD", "korean" : 90, "math" : 53, "english" : 78, "science" : 35}
]
students
```

```python
print("이름", "총점", "평균", sep = "\t")
for student in students:
    # 개별 학생에 대해서 무엇을 할지?
    score_sum = student["korean"] + student["math"] + student["english"] + student["science"]
    score_avg = score_sum/4
    print(student["name"], score_sum, score_avg, sep= "\t")
```

```python
# 하지만, 개별 학생들에 대해서도 요청한다면?
# 0번 학생 나와서 이름하고 4과목 성적 얘기해봐!
students[0]
```

```python
# 하지만... 학생들에 대한 정보를 일일히 입력하고 그러기 귀찮으니까
# 함수를 만들어줄까?
# 학생들마다 정보는 name, kor, math, eng, sci 로 해야지
# 매 학생마다 key값을 계속 동일하게 작성을 하니 귀찮아서..반복 돌려야하지 않겠나
```
