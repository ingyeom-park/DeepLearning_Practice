> 차원  

차원은 축의 갯수이다.  
0차원: 스칼라  
1차원: 벡터  
2차원: 행렬  
n차원: 고차원 텐서  
  
> 모양  

모양은 각 축의 길이이다.  
2차원이라도 `(3, 4)`인지, `(100, 10)` 인지에 따라 의미가 완전히 달라진다.  
딥러닝 모델은 입력이 `특정 유형 혹은 이런 모양`임을 강하게 강제함.  
그래서 모양이 한 칸만 어긋나도 바로 에러가 날 수 있음.

```python
# ndarray를 만들기 위한 준비
import numpy as np
```

```python
arr = np.array(10) #숫자 10 하나를 담은 배열, 축이 없음
print(arr, arr.ndim, arr.shape)
# 10 0 ()
```

```python
arr = np.array([10])
print(arr, arr.ndim, arr.shape)
# [10] 1 (1,)
```

```python
arr = np.array([100, 200])
print(arr, arr.ndim, arr.shape)
# 첫 번째 축의 길이가 2다 (2,)
```

2차원 Matrix는 가로와 세로 길이가 딱 맞아야 합니다.  
행렬은 “각 행이 같은 길이”를 가져야 합니다.   
한 행은 원소가 3개인데 다른 행은 원소가 2개이면, 수학적으로도 깔끔한 2차원 행렬로 보기 어렵고, 딥러닝 프레임워크도 그런 입력을 기본 형태로 받아들이지 못합니다.  

numpy 자료형은 원소 dtype까지 통일되는 성격이 있습니다.  
numpy 배열은 내부적으로 한 덩어리 메모리에 같은 타입으로 저장하는 구조라서, 원소 타입이 섞이면 자동으로 더 큰 범주로 변환하거나(예: int와 float가 섞이면 float로), 심하면 object로 떨어질 수 있습니다.   
그래서 “자료형까지 통일”이라는 말은 실습에서는 보통 “숫자 배열은 숫자 dtype로 깨끗하게 유지하자”는 의미로 읽으시면 됩니다.  

파이썬 리스트로 2차원을 간접적으로 표현해 쓰는 것은 우회입니다.  
[[...], [...]] 형태는 “리스트 안에 리스트”로 2차원을 흉내 낸 것입니다.   
이 구조는 편하긴 하지만, 각 행 길이가 달라도 파이썬은 막지 않습니다.   
그래서 “줄을 잘 맞춰서, 가로 세로 인덱스로, 정수 인덱스 기반으로 우회적 접근”이라는 표현이 나옵니다.   
즉, 사람이 규칙을 지켜야만 2차원처럼 쓸 수 있다는 뜻입니다.  

```python
m = [[1, 2, 3], [4, 5, 6]]
m
```

```python
arr = np.array(m)
print(arr.ndim) # 축이 2개라서 2차원
print(arr.shape)
```
