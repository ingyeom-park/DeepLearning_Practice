```python
# 참고) 외부에서 내부의 속성(Property)에 접근할 수도 있습니다.
class Square:
    # 정사각형을 담당하는 클래스
    
    def __init__ (self, size):
        self.size = size

    def get_area(self):
        return self.size * self.size
```

```python
# 구체적인 정사각형 1개를 생성해보자. (대충 길이는 10으로)

s1 = Square(10)
print(s1.size)
print(s1.get_area())
```

```python
s2 = Square(20)
print(s2.size)
print(s2.get_area())
```

```python
# 생성된 이후에 외부에서 내부의 속성값을 변경하는 것도 가능함.
s1.size = 30
s1.get_area()
```

```python
# 혹은 외부에서의 접근을 차단할 수도 있을듯?
# 속성값에 대한 접근에 대해서 요청: get, 변경에 대한 요청: set
# 그리고 언더바 2개로 입력을 막을 숭 ㅣㅆ음.

class Square:
    def __init__(self, size):
        self.__size = size

    def get_area(self):
        return self.__size * self.__size
    
    # 내부 속성 값에 대한 정보 전달; get
    def get_size(self):
        return self.__size
    
    # 내부 속성 값에 대한 변경
    def set_size(self, value):
        self.__size = value
```

```python
class Student:
    def __init__(self, name, kor, math, eng, sci):
        self.name = name
        self.korean = kor
        self.math = math
        self.english = eng
        self.science = sci

    # 이제, 학생이라면 해야하는 기능을 추가하려고 함 (총점계산, 평균계산, 출력 등)
    # 내가 필요한 기능/함수들을 설계하고 메서드도 설계하고

    
    def get_sum (self):
        scores_sum = self.korean + self.math + self.english + self.science
        return scores_sum

    def get_avg (self):
        score_avg = self.get_sum() / 4
        return score_avg
    
    def to_string(self):
        temp = "{0}\t{1}\t{2}".format(self.name, self.get_sum(), self.get_avg())
        return temp
    
    # 개인 취향이긴 한데,
    # 2명 학생들에 대해서 비교하는 것을 객체 기반 연산자로 해볼까?
    # 기준은 4과목 총점으로 하겠음. 
    def __eq__ (self, value): #std0 == std3 , value, 클래스로 생성된 객체
        return self.get_sum() == value.get_sum()
```

```python
students = [
    Student("AAA", 98, 93, 29 ,34)
]
```

```python
print(
    students[0].to_string()
)
```
